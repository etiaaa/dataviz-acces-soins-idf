<h2 style="border-left: 6px solid #1e40af; padding-left: 18px; margin: 36px 0 20px 0; font-weight: 800; font-size: 1.6em; line-height: 1.2;"><span style="font-weight: 800; margin-right: 10px;">1.</span>Le projet</h2>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Trouver un médecin en France, c'est de plus en plus dur. Pour un généraliste, le délai d'attente a <strong>triplé en 7 ans</strong> (12 jours en 2026 contre 4 en 2019). Côté spécialistes, l'étude <strong>Doctolib &times; Fondation Jean-Jaurès 2026</strong> (234 millions de consultations) chiffre <strong>42 jours</strong> d'attente moyenne pour un cardiologue &mdash; et jusqu'à plus de 5 mois dans les départements les plus tendus &mdash; et plus d'un mois pour un dermatologue. <strong>84 % des Français</strong> disent avoir eu des difficultés à obtenir un RDV chez un spécialiste en 2025.</p>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Les <strong>élus locaux</strong> &mdash; maires et présidents d'intercommunalité (EPCI) &mdash; ont des leviers concrets pour agir : ouvrir une maison de santé, salarier un médecin, installer une cabine de télémédecine. Mais aujourd'hui ils n'ont pas d'outil simple pour décider <strong>où agir en priorité</strong> sur leur territoire.</p>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Mon dashboard répond à ça : il donne aux élus locaux une vue claire de leur territoire pour <strong>prioriser les communes les plus en tension</strong> et choisir le bon levier d'action.</p>

<h2 style="border-left: 6px solid #1e40af; padding-left: 18px; margin: 36px 0 20px 0; font-weight: 800; font-size: 1.6em; line-height: 1.2;"><span style="font-weight: 800; margin-right: 10px;">2.</span>Pourquoi ces choix ?</h2>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Au départ je voulais juste faire un dashboard sur l'accès aux soins. Mais en cadrant le sujet, j'ai dû <strong>changer plusieurs fois d'avis</strong> &mdash; et la disponibilité réelle des données m'a obligée à en revoir certains. Voici les principaux choix et pourquoi je les ai pris.</p>

| Question | Idée de départ | Choix final | Pourquoi |
| :--- | :--- | :--- | :--- |
| **À qui s'adresse le dashboard ?** | Citoyens | **Élus locaux** (maires + présidents d'EPCI) | Les citoyens ne peuvent que consommer l'offre, pas la créer. Les élus ont les vrais leviers (MSP, salariat, télémédecine). |
| **Quels médecins ?** | Généralistes + pédiatre, dermato, gynéco, cardio (1 par public) | **Généraliste + 4 spécialités les plus tendues** : cardiologie, dermatologie, ophtalmologie, gynécologie | Critère revu : non plus « 1 spé par public » mais **les spécialités où il faut vraiment ramer** (délais Doctolib–Jean-Jaurès 2026). La pédiatrie, peu tendue en délais, sort au profit de l'ophtalmologie. |
| **Granularité des spécialistes ?** | Commune (je pensais que Cartosanté la donnait) | **Département** | Vérification faite : Cartosanté et l'APL DREES ne couvrent que le **premier recours** (généralistes, dentistes, kinés…). Ces 4 spécialités ne sont en open data qu'au **département** (RPPS/DREES). Je les rattache au département, diffusé vers les communes. |
| **Quel angle d'analyse ?** | « Où les délais sont les plus longs ? » | **« Où l'offre est la plus tendue ? »** | Les délais par commune n'existent pas en open data. L'offre, si (APL, densité, âge). On travaille avec ce qu'on a vraiment. |
| **Seuil départs en retraite ?** | % médecins > 55 ans | **Âge moyen + % > 60 ans** | 55 ans c'est jeune pour un médecin. 60 ans = vrai horizon de départ à 5–10 ans. |
| **Portée géographique ?** | France entière | **Île-de-France** | 3 semaines : mieux vaut approfondir 1 région. Le contraste Paris / Seine-Saint-Denis raconte déjà une histoire forte. |

<div style="background: #eff6ff; border-left: 4px solid #3b82f6; padding: 14px 20px; border-radius: 6px; margin: 18px 0;"><div style="color: #1e40af; text-transform: uppercase; letter-spacing: 0.06em; font-size: 0.78em; font-weight: 700; margin-bottom: 6px;">J'ai aussi challengé mes propres chiffres</div><div style="color: #0f172a;">J'ai écrit « dermato à 6 mois » sans source au départ. Première vérif : la DREES annonçait 61 jours en 2018 &rarr; j'ai cru avoir exagéré. Deuxième vérif en réalisant qu'on est en <strong>2026, pas 2018</strong> : aujourd'hui c'est <strong>plus d'un mois en moyenne</strong> pour un dermatologue, et plusieurs mois dans les zones tendues. La situation s'est dégradée bien plus que je ne le pensais. <strong>Toujours dater ses sources.</strong></div></div>

<h2 style="border-left: 6px solid #1e40af; padding-left: 18px; margin: 36px 0 20px 0; font-weight: 800; font-size: 1.6em; line-height: 1.2;"><span style="font-weight: 800; margin-right: 10px;">3.</span>Solutions pour les élus</h2>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Ce que les élus peuvent <strong>faire</strong> avec le dashboard.</p>

- **Subventionner l'installation** d'un médecin (aide à l'installation, logement, équipement)
- **Ouvrir une maison de santé pluridisciplinaire** (MSP) — attire mieux qu'un cabinet seul
- **Salarier un médecin** (modèle de plus en plus répandu pour les communes)
- **Installer une cabine de télémédecine** quand pas de médecin sur place
- **Mutualiser en intercommunalité** (partager un médecin / un projet entre plusieurs communes)

<div style="background: #fffbeb; border-left: 4px solid #f59e0b; padding: 14px 20px; border-radius: 6px; margin: 18px 0;"><div style="color: #b45309; text-transform: uppercase; letter-spacing: 0.06em; font-size: 0.78em; font-weight: 700; margin-bottom: 6px;">Risque de bouchons — comment je l'évite</div><div style="color: #0f172a;">Si le dashboard montre un classement « pire / meilleur », tout le monde concentre ses efforts au même endroit. Je préfère afficher la position relative dans l'EPCI (= « parmi les communes voisines, lesquelles sont en tension »), pas un classement national. Le but est de <strong>décider localement</strong>, pas de comparer toutes les communes de France.</div></div>


<h2 style="border-left: 6px solid #1e40af; padding-left: 18px; margin: 36px 0 20px 0; font-weight: 800; font-size: 1.6em; line-height: 1.2;"><span style="font-weight: 800; margin-right: 10px;">4.</span>Mes KPIs</h2>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Trois indicateurs, à deux niveaux de granularité.</p>

| KPI | Granularité | Source |
| :--- | :--- | :--- |
| **APL généralistes** | Commune | DREES |
| **Densité de 4 spécialistes** (cardiologie, dermatologie, ophtalmologie, gynécologie) | Département | RPPS / DREES |
| **Âge moyen des médecins** | Département | data.drees (RPPS, 2025) |

<div style="background: #eff6ff; border-left: 4px solid #3b82f6; padding: 14px 20px; border-radius: 6px; margin: 18px 0;"><div style="color: #1e40af; text-transform: uppercase; letter-spacing: 0.06em; font-size: 0.78em; font-weight: 700; margin-bottom: 6px;">Score composite « tension d'accès »</div><div style="color: #0f172a;">Je combine les <strong>3 KPIs</strong> en un <strong>score 0–100</strong> par commune, <strong>⅓ chacun</strong> (APL généralistes, densité spécialistes, âge des médecins). Normalisation min-max winsorisée (percentiles 2/98) puis, pour l'APL et la densité, inversion (peu d'offre = forte tension) ; pour l'âge, sens direct (médecins âgés = risque de départs = tension). J'affiche aussi un <strong>rang relatif intra-EPCI</strong> pour décider localement.</div></div>

<h2 style="border-left: 6px solid #1e40af; padding-left: 18px; margin: 36px 0 20px 0; font-weight: 800; font-size: 1.6em; line-height: 1.2;"><span style="font-weight: 800; margin-right: 10px;">5.</span>Sources de données</h2>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;">Toutes <strong>publiques et gratuites</strong>.</p>

- **DREES** — APL des soins de premier recours (généralistes), à la commune
- **RPPS / DREES** — effectifs de médecins spécialistes libéraux, par département
- **CNOM** — Atlas de la démographie médicale 2026 (âge des médecins) — *document PDF, KPI à intégrer ; alternative exploitable : data.drees*
- **INSEE** — Code Officiel Géographique + **BANATIC** (correspondance commune ↔ EPCI) + population municipale
- **Contours des communes** (GeoJSON, IGN / France-GeoJSON) — fond de carte pour Tableau

<div style="background: #fffbeb; border-left: 4px solid #f59e0b; padding: 14px 20px; border-radius: 6px; margin: 18px 0;"><div style="color: #b45309; text-transform: uppercase; letter-spacing: 0.06em; font-size: 0.78em; font-weight: 700; margin-bottom: 6px;">Limites assumées</div><div style="color: #0f172a;">L'open data ne fournit pas les <strong>délais</strong> de RDV par commune (DREES : enquête nationale 2018). Je travaille donc sur l'<strong>offre médicale</strong> (APL, densité, âge), pas sur l'attente ; les délais 2026 servent à justifier l'urgence en intro. De plus, la densité des spécialistes n'existe qu'au <strong>département</strong> : à l'intérieur d'un département elle est constante, la finesse à la commune vient des généralistes.</div></div>

<p style="font-size: 1.08em; line-height: 1.55; margin: 14px 0;"><strong>Deux vues complémentaires</strong> dans Tableau Public, publiées en ligne.</p>

- **Vue EPCI (président d'intercommunalité)** : carte de l'Île-de-France, communes coloriées par **score de tension** (vert = bien doté, rouge = tendu), avec un **filtre par intercommunalité**. Permet de repérer **où mutualiser en priorité** (MSP, télémédecine, coopération…).
